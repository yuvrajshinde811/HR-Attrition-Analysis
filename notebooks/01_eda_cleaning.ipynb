```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv(r'C:\Users\YUVRAJ\Desktop\My Activities\Projects\hr_attrition_project\hr_attrition.csv')

print(df.shape)          # how many rows and columns?
print(df.dtypes)         # what type is each column?
print(df.isnull().sum()) # any missing values?
print(df.duplicated().sum())  # duplicate rows?
df.head()
```

```python
df = df.drop_duplicates()

df = df.drop(columns=['EmployeeCount','StandardHours',
                       'Over18'],
            errors='ignore'
            )

df['Attrition'] = df['Attrition'].map({'Yes':1, 'No':0})

import os
os.makedirs('../outputs', exist_ok=True)

df.to_csv('../outputs/cleaned_data.csv', index=False)
print("Cleaned. Shape:", df.shape)
```

```python
df['TenureBucket'] = pd.cut(df['YearsAtCompany'],
    bins=[-1, 2, 5, 10, 40],
    labels=['0-2 yrs', '3-5 yrs', '6-10 yrs', '10+ yrs'])

df['SalaryBand'] = pd.cut(df['MonthlyIncome'],
    bins=[0, 3000, 6000, 10000, 25000],
    labels=['Low', 'Medium', 'High', 'Very High'])

df['HighRiskFlag'] = (
    (df['OverTime'] == 'Yes') &
    (df['JobSatisfaction'] <= 2) &
    (df['YearsAtCompany'] <= 3)
).astype(int)

print(df[['TenureBucket', 'SalaryBand', 'HighRiskFlag']].head(10))
print("\nHigh risk employees:", df['HighRiskFlag'].sum())

df.to_csv('../outputs/cleaned_data.csv', index=False)
print("Saved. Final shape:", df.shape)
```
